```python
#중복되고 흐린 이미지 삭제

# 📦 필요한 라이브러리 설치
!pip install imagehash opencv-python-headless

import os
import zipfile
import shutil
from PIL import Image
import imagehash
import cv2
import numpy as np

# 📁 경로 설정
uploaded_zip = 'images.zip'  # 업로드할 zip 파일 이름
extract_dir = '/content/extracted_images'
output_dir = '/content/filtered_images'
final_zip_path = '/content/cleaned_images.zip'

# 1. 📂 디렉토리 준비
shutil.rmtree(extract_dir, ignore_errors=True)
shutil.rmtree(output_dir, ignore_errors=True)
os.makedirs(extract_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

# 2. 🔓 압축 해제
with zipfile.ZipFile(uploaded_zip, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# 3. 🔍 흐림 이미지 판별 함수
def is_blurry(img, threshold=100.0):
    gray = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2GRAY)
    var = cv2.Laplacian(gray, cv2.CV_64F).var()
    return var < threshold

# 4. 🧹 중복 제거 + 흐림 제거
hashes = set()
kept_count = 0

for root, _, files in os.walk(extract_dir):
    for fname in files:
        fpath = os.path.join(root, fname)
        try:
            with Image.open(fpath) as img:
                img = img.convert("RGB")
                h = imagehash.average_hash(img)

                if h in hashes or is_blurry(img):
                    continue

                hashes.add(h)
                img.save(os.path.join(output_dir, fname))
                kept_count += 1
        except:
            continue

print(f"✅ 정리된 이미지 수: {kept_count}")

# 5. 📦 다시 zip 압축
shutil.make_archive(output_dir, 'zip', output_dir)
shutil.move(f"{output_dir}.zip", final_zip_path)

print(f"📁 최종 압축 파일 경로: {final_zip_path}")
```

```python
#비문이 보이는 코드만 나오게 한 느낌

!pip install opencv-python-headless

import os
import zipfile
import shutil
import cv2
import numpy as np
from PIL import Image
from google.colab import files

# 경로 설정
uploaded_zip = 'images.zip'
extract_dir = '/content/extracted_images1'
filtered_dir = '/content/nose_visible_images1'
final_zip_path = '/content/nose_filtered1.zip'

# 디렉토리 정리
shutil.rmtree(extract_dir, ignore_errors=True)
shutil.rmtree(filtered_dir, ignore_errors=True)
os.makedirs(extract_dir, exist_ok=True)
os.makedirs(filtered_dir, exist_ok=True)

# 압축 해제
with zipfile.ZipFile(uploaded_zip, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# 흐림 판단 함수
def is_blurry(img, threshold=100.0):
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    var = cv2.Laplacian(gray, cv2.CV_64F).var()
    return var < threshold

# 비문 존재 여부 판단 (하단 중앙에 어두운 원형/타원 검출)
def has_nose_like_region(img):
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    h, w = gray.shape
    roi = gray[h//2:, w//4:3*w//4]  # 하단 중앙 영역

    _, thresh = cv2.threshold(roi, 70, 255, cv2.THRESH_BINARY_INV)
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area > 500:  # 너무 작은 건 무시
            (x, y), radius = cv2.minEnclosingCircle(cnt)
            circularity = 4 * np.pi * area / (cv2.arcLength(cnt, True) ** 2 + 1e-6)
            if 0.5 < circularity < 1.3:
                return True
    return False

# 이미지 필터링
kept_count = 0
for root, _, files_in_dir in os.walk(extract_dir):
    for fname in files_in_dir:
        fpath = os.path.join(root, fname)
        try:
            img = Image.open(fpath).convert("RGB")
            img_cv = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)

            if is_blurry(img_cv):
                continue
            if not has_nose_like_region(img_cv):
                continue

            img.save(os.path.join(filtered_dir, fname))
            kept_count += 1
        except:
            continue

print(f"✅ 비문이 보이는 이미지 수: {kept_count}")

# 압축
shutil.make_archive(filtered_dir, 'zip', filtered_dir)
shutil.move(f"{filtered_dir}.zip", final_zip_path)

print(f"📁 최종 압축 파일: {final_zip_path}")
```
