```python
import yaml

nose_data = {'train' : '/content/nose_data/train/images/',
        'test' : '/content/nose_data/test/images',
        'val' : '/content/nose_data/val/images/',
        'names' : ['nose'],
        'nc': 1 }

# nose_data.yaml 파일 생성
with open('/content/nose_data/data.yaml', 'w') as f:
    yaml.dump(nose_data, f)

with open('/content/nose_data/data.yaml', 'r') as f:
    print(yaml.safe_load(f))
```

```python
import ultralytics
from ultralytics import YOLO

model = YOLO('yolo11n.pt')
```

```python
model.train(data='/content/nose_data/data.yaml', epochs=30, patience=5, imgsz=640)
```

```python
result = model(source='/content/test/images', save=True)
```
