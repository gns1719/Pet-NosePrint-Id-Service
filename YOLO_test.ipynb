```python
import yaml
data = {'train' : '/content/nose_data/train/images/',
        'test' : '/content/nose_data/test/images',
        'val' : '/content/nose_data/val/images/',
        'names' : ['개코'],
        'nc': 1 }

with open('/content/nose_data/data.yaml', 'w') as f:
  yaml.dump(data, f)

with open('/content/nose_data/data.yaml', 'r') as f:
  display(yaml.safe_load(f))
```

```python
import ultralytics
from ultralytics import YOLO

model = YOLO('yolo11n.pt')
```

```python
print(type(model.names), len(model.names))

print(model.names)
```

```python
model.train(data='/content/nose_data/data.yaml', epochs=30, patience=5, imgsz=640)
```

```python
print(type(model.names), len(model.names))

print(model.names)
```

```python
result = model(source='/content/nose_data/test/images', save=True)
```
